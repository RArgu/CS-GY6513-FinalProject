# Phase 3 - Spike Detection

## Overview

This notebook detects anomalous price movements in the tick data that may signal insider trading activity ahead of known resolution events

### Pipeline

1. We read `polymarket.ticks` from BigQuery
2. Then we aggregated trades into **5 minute VWAP buckets** per market and side
3. We write that to `polymarket.price_aggregates`
4. Then we compute **rolling mean and standard deviation** over the last 20 buckets
5. Compute a **z-score** for each bucket's VWAP
6. Flag buckets where z-score > threshold as spike events
7. Write the flagged events to `polymarket.spike_events`
8. We crosscheck the spikes against known resolution timestamps

## 1. Setup

### 1.1 Imports and Constants

```python
%matplotlib inline
import time
from pathlib import Path

from pyspark.sql import SparkSession, Window
from pyspark.sql import functions as F

# GCP config
PROJECT = "sr8250-cs6513-polymarket"
DATASET = "polymarket"
BUCKET = "cs6513-polymarket"

ROOT = Path.cwd().parent
JAR = str(ROOT / "jars" / "spark-bigquery-with-dependencies_2.12-0.36.1.jar")

SPARK_MASTER = "local[*]"

# Detection parameters
BUCKET_DURATION = "5 minutes"  # VWAP window size
WINDOW_SIZE = 20  # rolling baseline: 20 preceding buckets
Z_THRESHOLD = 3.0  # flag bucket if z-score exceeds this
MIN_VOLUME = 100.0  # drop buckets below this USD volume
MIN_STD = 0.002    # ignore z-scores computed against a near-flat baseline (floating-point artifact)
```

### 1.2 Spark Session

```python
spark = (
    SparkSession.builder
    .appName("polymarket-spike-detection")
    .master(SPARK_MASTER)
    .config("spark.jars", JAR)
    .config("spark.driver.host", "localhost")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")

print(f"App name = {spark.sparkContext.appName}")
```

### 1.3 Helper function for writing Dataframe to BigQuery

```python
def write_bq(df, table):
    """
    Writes a Spark DataFrame to a BigQuery table, overwriting any existing data

    Args:
        df: The Spark DataFrame to write
        table: The BigQuery table name within the configured dataset
    """
    # 2 partitions = 2 concurrent write streams; more overwhelms the Storage Write API
    (
        df.repartition(2)
        .write
        .format("bigquery")
        .option("table", f"{PROJECT}.{DATASET}.{table}")
        .option("writeMethod", "direct")
        .mode("overwrite")
        .save()
    )
```

## 2. Load Tick Data from BigQuery

```python
t0 = time.time()

ticks = (
    spark.read
    .format("bigquery")
    .option("table", f"{PROJECT}.{DATASET}.ticks")
    .load()
)

tick_count = ticks.count()
elapsed = time.time() - t0

print(f"Ticks loaded : {tick_count:,}")
print(f"Time         : {elapsed:.1f}s")
ticks.printSchema()
```

```python
# row counts per market
ticks.groupBy("market_slug", "side").count().orderBy("market_slug", "side").show(truncate=False)
```

> **What this means**
> 
> **329,373 ticks** were loaded from the `polymarket.ticks` BigQuery table. Each tick is one individual trade — every time someone bought > or sold a YES or NO contract on any of our 6 markets that is one row here.

## 3. Computeing 5 Minute VWAP Buckets

**VWAP (Volume-Weighted Average Price)** weights each trade by its dollar size. We have used weighted avg. so that a single large fill does not distort the bucket price 

$$\text{VWAP} = \frac{\sum (\text{price}_i \times \text{size}_i)}{\sum \text{size}_i}$$

Each bucket is keyed on `(market_slug, side, window_start)`.

```python
t0 = time.time()

buckets = (
    ticks
    .groupBy(
        "market_slug",
        "side",
        F.window("timestamp", BUCKET_DURATION).alias("w"),
    )
    .agg(
        (F.sum(F.col("implied_price") * F.col("size_usd")) / F.sum("size_usd")).alias("vwap"),
        F.sum("size_usd").alias("volume_usd"),
        F.count("*").alias("trade_count"),
    )
    .select(
        "market_slug",
        "side",
        F.col("w.start").alias("window_start"),
        F.col("w.end").alias("window_end"),
        "vwap",
        "volume_usd",
        "trade_count",
    )
    .orderBy("market_slug", "side", "window_start")
)

bucket_count = buckets.count()
elapsed = time.time() - t0

print(f"5-minute buckets : {bucket_count:,}")
print(f"Time             : {elapsed:.1f}s")
buckets.show(5, truncate=False)
```

> **What this means:**
>
> We compressed 329,373 individual ticks into **65,985  5 minute buckets**. So now instead of looking at every single trade we now have one summary row per 5 minute window per market per side.
>
> - `window_start` / `window_end` — the start and end of the 5-minute window (e.g. 22:20 to 22:25)
> - `vwap` — the volume-weighted average price in that window (0.75 means YES contracts were trading at 75 cents = 75% implied probability)
> - `volume_usd` — total dollars traded in that 5-minute period (24.64 here is very low — thin overnight trading)
> - `trade_count` — number of individual fills in the window (1 here means just one trade happened)

Writing aggregates to `price_aggregates` so later on in phases like visualization, bet tracing we can query pre computed VWAP directly without re running aggregation

```python
t0 = time.time()
write_bq(buckets, "price_aggregates")
print(f"price_aggregates written ({time.time() - t0:.1f}s)")
```

## 4. Rolling Statistics and Z-Score

For each bucket we compute a **rolling baseline** from the 20 preceding buckets and express the current VWAP as a z-score relative to that baseline

$$z = \frac{\text{VWAP} - \mu_{\text{rolling}}}{\sigma_{\text{rolling}}}$$

**Key design choices:**
- The rolling window uses `rowsBetween(-20, -1)` — **excluding the current row**. We dont include the current row because including it would pull the mean toward the spike and can artificially reduce the z-score.
- Buckets 3 to 20 produce z-scores using an incomplete window. The window is fully stable from bucket 21 onwards
- Z-score is set to `NULL` when `rolling_std = 0` (price was perfectly flat). Flagging those as spikes would be a false positive.

```python
t0 = time.time()

# exclude current row so the baseline is purely historical
roll_w = (
    Window
    .partitionBy("market_slug", "side")
    .orderBy("window_start")
    .rowsBetween(-WINDOW_SIZE, -1)
)

buckets = (
    buckets
    .withColumn("rolling_mean", F.mean("vwap").over(roll_w))
    .withColumn("rolling_std", F.stddev("vwap").over(roll_w))
    .withColumn(
        "z_score",
        F.when(
            F.col("rolling_std") >= MIN_STD,
            (F.col("vwap") - F.col("rolling_mean")) / F.col("rolling_std"),
        ).otherwise(F.lit(None))
    )
)

buckets.select(
    "market_slug", "side", "window_start", "vwap",
    "rolling_mean", "rolling_std", "z_score"
).orderBy("market_slug", "side", "window_start").show(10, truncate=False)

print(f"Time taken {time.time() - t0:.1f}s")
```

## Visualisation for VWAP Price Over Time

Chart per resolved market showing how the price evolved and a sharp jump toward the resolution line is the spike we are trying to detect.

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd

# convert to pandas for plotting
buckets_pd = buckets.select(
    "market_slug", "side", "window_start", "vwap", "volume_usd"
).toPandas()
buckets_pd["window_start"] = buckets_pd["window_start"].dt.tz_localize(None)

all_markets = [
    "us-strikes-iran-feb28",
    "khamenei-out-feb28",
    "maduro-out-jan31",
    "us-forces-iran-mar31",
    "us-iran-ceasefire-jun30",
    "china-invades-taiwan-2026",
]
resolution_dates = {
    "us-strikes-iran-feb28": "2026-02-28",
    "khamenei-out-feb28":    "2026-02-28",
    "maduro-out-jan31":      "2026-01-03",
}
market_labels = {
    "us-strikes-iran-feb28":     "US Strikes Iran (resolved YES)",
    "khamenei-out-feb28":        "Khamenei Out (resolved YES)",
    "maduro-out-jan31":          "Maduro Out (resolved YES)",
    "us-forces-iran-mar31":      "US Forces Iran (active)",
    "us-iran-ceasefire-jun30":   "US-Iran Ceasefire (active)",
    "china-invades-taiwan-2026": "China Invades Taiwan (active)",
}

for slug in all_markets:
    fig, ax = plt.subplots(figsize=(14, 4))
    mdf = buckets_pd[buckets_pd["market_slug"] == slug]
    for side, color, label in [("yes", "#2196F3", "YES"), ("no", "#F44336", "NO")]:
        sdf = mdf[mdf["side"] == side].sort_values("window_start")
        ax.plot(sdf["window_start"], sdf["vwap"], color=color, linewidth=0.8, label=label)
    if slug in resolution_dates:
        ax.axvline(x=pd.Timestamp(resolution_dates[slug]), color="black", linestyle="--", linewidth=1.2, label="Resolution")
    ax.set_title(f"VWAP Price Over Time — {market_labels[slug]}", fontsize=12, fontweight="bold")
    ax.set_ylabel("VWAP (implied price)")
    ax.set_ylim(0, 1.05)
    ax.legend(loc="upper left", fontsize=9)
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%b %d"))
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(f"charts/vwap_{slug}.png", dpi=300, bbox_inches="tight")
    plt.show()
```

### Adding `price_before` and `price_after`

Using `lag` and `lead` window functions to pull the VWAP of the immediately adjacent buckets. These fields let us quantify how sharp the price jump was like price went from 0.08 to 0.74 in one 5 minute window.

```python
lag_w = (
    Window
    .partitionBy("market_slug", "side")
    .orderBy("window_start")
)

buckets = (
    buckets
    .withColumn("price_before", F.lag("vwap", 1).over(lag_w))
    .withColumn("price_after", F.lead("vwap", 1).over(lag_w))
)
```

## 5. Flag Spike Events

A bucket is flagged as a spike if:
1. `z_score > Z_THRESHOLD` (anomalous price move relative to baseline)
2. `volume_usd >= MIN_VOLUME` (at least $100 traded filters single-fill artifacts in thin periods)
3. `z_score is not NULL` (so that initial warmup buckets excluded)

```python
t0 = time.time()

spikes = (
    buckets
    .filter(F.col("volume_usd") >= MIN_VOLUME)
    .filter(F.col("z_score").isNotNull())
    .filter(F.col("z_score") > Z_THRESHOLD)
    .select(
        "market_slug",
        "side",
        "window_start",
        "window_end",
        "vwap",
        "price_before",
        "price_after",
        "rolling_mean",
        "rolling_std",
        "z_score",
        "volume_usd",
        "trade_count",
    )
    .orderBy("market_slug", "side", "window_start")
)

spike_count = spikes.count()
elapsed = time.time() - t0

print(f"Spike events detected : {spike_count}")
print(f"Z threshold           : {Z_THRESHOLD}")
print(f"Min volume (USD)      : ${MIN_VOLUME:,.0f}")
print(f"Time                  : {elapsed:.1f}s")
```

> **What this output means:**
>
> **569 spike events** were detected across all 6 markets. A spike event is a 5-minute bucket where:
> 1. The price moved more than **3 standard deviations** above the recent rolling baseline
> 2. At least **$100** was traded in that window
>
> 311 out of 65985 total buckets so roughly about **0.86% of all buckets** were flagged as spikes

## 6. Sanity Check for Resolved Markets

For the resolved markets we know exactly when the price spike occurred so we verify that the detector fired at the right time.

| Market | Expected spike |
|---|---|
| `us-strikes-iran-feb28` | ~2026-02-28|
| `khamenei-out-feb28` | ~2026-02-28 |

## Parameter Experiments

### Experiment - Z-Score Threshold Sensitivity

How spike count changes as we raise the threshold. Our goal here was to find the value where resolved markets still show spikes near resolution but active markets stop flooding with false positives

```python
thresholds = [2.0, 2.5, 3.0, 3.5, 4.0, 5.0]
all_markets = [
    "us-strikes-iran-feb28", "khamenei-out-feb28", "maduro-out-jan31",
    "us-forces-iran-mar31", "us-iran-ceasefire-jun30", "china-invades-taiwan-2026"
]

print(f"{'Market':<30} {'z=2.0':>6} {'z=2.5':>6} {'z=3.0':>6} {'z=3.5':>6} {'z=4.0':>6} {'z=5.0':>6}")
print("-" * 72)

rows = {}
for slug in all_markets:
    counts = []
    for z in thresholds:
        n = (
            buckets
            .filter(F.col("market_slug") == slug)
            .filter(F.col("volume_usd") >= MIN_VOLUME)
            .filter(F.col("z_score").isNotNull())
            .filter(F.col("z_score") > z)
            .count()
        )
        counts.append(n)
    rows[slug] = counts
    short = slug[:28]
    print(f"{short:<30} {counts[0]:>6} {counts[1]:>6} {counts[2]:>6} {counts[3]:>6} {counts[4]:>6} {counts[5]:>6}")

# bar chart
import numpy as np
x = np.arange(len(all_markets))
width = 0.13
fig, ax = plt.subplots(figsize=(14, 6))
for i, (z, color) in enumerate(zip(thresholds, ["#EF5350","#FF7043","#FFA726","#66BB6A","#42A5F5","#7E57C2"])):
    vals = [rows[s][i] for s in all_markets]
    ax.bar(x + i * width, vals, width, label=f"z={z}", color=color, alpha=0.85)
ax.set_xticks(x + width * 2.5)
ax.set_xticklabels([s.replace("-", "-\n") for s in all_markets], fontsize=8)
ax.set_ylabel("Spike count")
ax.set_title("Spike Count vs Z-Score Threshold per Market")
ax.legend(title="Threshold")
ax.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("charts/threshold_sensitivity.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: charts/threshold_sensitivity.png")
```

> **What this means:**
>
> - At **z=2.0** spike counts are high and too many borderline movements are flagged which will make it hard to separate signal from noise.
> - At **z=3.0** counts drop significantly but the resolution events are still captured for all 3 resolved markets.
> - At **z=4.0** counts halve again. The main resolution spikes are still there but smaller pre event accumulation signals may get filtered out
> - At **z=5.0** the counts are very low and we may be lose early pre event signals.

### Experiment — Window Size Sensitivity

- How does changing the rolling window size affect the z-score ? We did this experiement with only 1 market for testing  at the Maduro resolution spike. 
- Larger window = more stable baseline while smaller = reacts faster but noisier.

```python
window_sizes = [5, 10, 20, 30, 50]
target_slug = "maduro-out-jan31"
target_ts   = "2026-01-03"

print(f"{'WINDOW_SIZE':<15} {'Z-score at resolution':<25} {'Total spikes in market'}")
print("-" * 58)

ws_results = []
for ws in window_sizes:
    w = (
        Window
        .partitionBy("market_slug", "side")
        .orderBy("window_start")
        .rowsBetween(-ws, -1)
    )
    temp = (
        buckets
        .filter(F.col("market_slug") == target_slug)
        .withColumn("rm", F.mean("vwap").over(w))
        .withColumn("rs", F.stddev("vwap").over(w))
        .withColumn(
            "zs",
            F.when(F.col("rs") >= MIN_STD,
                   (F.col("vwap") - F.col("rm")) / F.col("rs")
            ).otherwise(F.lit(None))
        )
    )
    res_row = (
        temp
        .filter(F.col("side") == "yes")
        .filter(F.col("window_start").cast("date").cast("string") == target_ts)
        .agg(F.max("zs").alias("max_z"))
        .collect()
    )
    res_z = float(res_row[0]["max_z"]) if res_row and res_row[0]["max_z"] is not None else None
    total = (
        temp
        .filter(F.col("volume_usd") >= MIN_VOLUME)
        .filter(F.col("zs").isNotNull())
        .filter(F.col("zs") > Z_THRESHOLD)
        .count()
    )
    ws_results.append((ws, res_z or 0, total))
    z_str = f"{res_z:.1f}" if res_z else "NULL"
    print(f"{ws:<15} {z_str:<25} {total}")

# chart
fig, ax1 = plt.subplots(figsize=(8, 5))
ws_vals   = [r[0] for r in ws_results]
z_vals    = [r[1] for r in ws_results]
tot_vals  = [r[2] for r in ws_results]

ax1.plot(ws_vals, z_vals, "o-", color="#2196F3", linewidth=2, label="Z-score at resolution")
ax1.set_xlabel("Window size (buckets)")
ax1.set_ylabel("Z-score at resolution spike", color="#2196F3")
ax1.tick_params(axis="y", labelcolor="#2196F3")

ax2 = ax1.twinx()
ax2.bar(ws_vals, tot_vals, width=3, color="#F44336", alpha=0.4, label="Total spikes")
ax2.set_ylabel("Total spikes detected", color="#F44336")
ax2.tick_params(axis="y", labelcolor="#F44336")

ax1.set_title(f"Window Size Sensitivity — {target_slug} YES side")
ax1.grid(True, alpha=0.3)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right")
plt.tight_layout()
plt.savefig("charts/window_sensitivity.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: charts/window_sensitivity.png")
```
