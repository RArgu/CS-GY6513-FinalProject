## Overview

This notebook detects anomalous price movements in the tick data that may signal insider trading activity ahead of known resolution events.

### Pipeline

1. Read `polymarket.ticks` from BigQuery
2. Aggregate trades into **5-minute VWAP buckets** per market and side
3. Write `polymarket.price_aggregates` to BigQuery
4. Compute **rolling mean and standard deviation** over the last 20 buckets
5. Compute a **z-score** for each bucket's VWAP
6. Flag buckets where z-score > threshold as spike events
7. Write `polymarket.spike_events` to BigQuery
8. Sanity-check detected spikes against known resolution timestamps

### Execution Modes

Set `SPARK_MASTER` below to switch between local and standalone cluster:

| Mode | Value |
|---|---|
| Local (development) | `local[*]` |
| Standalone cluster | `spark://localhost:7077` |

```python
def write_bq(df, table):
    """
    Writes a Spark DataFrame to a BigQuery table, overwriting any existing data

    Args:
        df: The Spark DataFrame to write
        table: The BigQuery table name within the configured dataset
    """
    # 2 partitions = 2 concurrent write streams; more overwhelms the Storage Write API
    (
        df.repartition(2)
        .write
        .format("bigquery")
        .option("table", f"{PROJECT}.{DATASET}.{table}")
        .option("writeMethod", "direct")
        .mode("overwrite")
        .save()
    )
```

## 3. Compute 5-Minute VWAP Buckets

**VWAP (Volume-Weighted Average Price)** weights each trade by its dollar size:

$$\text{VWAP} = \frac{\sum (\text{price}_i \times \text{size}_i)}{\sum \text{size}_i}$$

Using VWAP instead of a simple average means a single large fill doesn't distort the bucket price. A 5-minute window is fine-grained enough to capture intraday spikes while smoothing individual tick noise.

Each bucket is keyed on `(market_slug, side, window_start)`.

```python
t0 = time.time()

buckets = (
    ticks
    .groupBy(
        "market_slug",
        "side",
        F.window("timestamp", BUCKET_DURATION).alias("w"),
    )
    .agg(
        (F.sum(F.col("implied_price") * F.col("size_usd")) / F.sum("size_usd")).alias("vwap"),
        F.sum("size_usd").alias("volume_usd"),
        F.count("*").alias("trade_count"),
    )
    .select(
        "market_slug",
        "side",
        F.col("w.start").alias("window_start"),
        F.col("w.end").alias("window_end"),
        "vwap",
        "volume_usd",
        "trade_count",
    )
    .orderBy("market_slug", "side", "window_start")
)

bucket_count = buckets.count()
elapsed = time.time() - t0

print(f"5-minute buckets : {bucket_count:,}")
print(f"Time             : {elapsed:.1f}s")
buckets.show(5, truncate=False)
```

> **What this output means:**
>
> The 329,373 individual ticks have been compressed into **65,985 five-minute buckets**. This is the core of the aggregation step — instead of looking at every single trade, we now have one summary row per 5-minute window per market per side.
>
> Each row in the sample shows:
> - `window_start` / `window_end` — the start and end of the 5-minute window (e.g. 22:20 to 22:25)
> - `vwap` — the volume-weighted average price in that window (0.75 means YES contracts were trading at 75 cents = 75% implied probability)
> - `volume_usd` — total dollars traded in that 5-minute period (24.64 here is very low — thin overnight trading)
> - `trade_count` — number of individual fills in the window (1 here means just one trade happened)
>
> The 65,985 buckets vs 329,373 ticks ratio means there were on average about **5 trades per 5-minute window** across all markets.

### Write `price_aggregates` to BigQuery

Persisting aggregates so later phases (visualization, bet tracing) can query pre-computed VWAP directly without re-running the aggregation.

```python
t0 = time.time()
write_bq(buckets, "price_aggregates")
print(f"price_aggregates written ({time.time() - t0:.1f}s)")
```

## 4. Rolling Statistics and Z-Score

For each bucket we compute a **rolling baseline** from the 20 preceding buckets and express the current VWAP as a z-score relative to that baseline:

$$z = \frac{\text{VWAP} - \mu_{\text{rolling}}}{\sigma_{\text{rolling}}}$$

**Key design choices:**
- The rolling window uses `rowsBetween(-20, -1)` — **excluding the current row**. Including it would pull the mean toward the spike and artificially reduce the z-score.
- The first 20 buckets per `(market_slug, side)` have no baseline yet and produce `z_score = NULL`. These are excluded when flagging spikes.
- Z-score is set to `NULL` when `rolling_std = 0` (price was perfectly flat). Flagging those as spikes would be a false positive.

```python
# exclude current row so the baseline is purely historical
roll_w = (
    Window
    .partitionBy("market_slug", "side")
    .orderBy("window_start")
    .rowsBetween(-WINDOW_SIZE, -1)
)

buckets = (
    buckets
    .withColumn("rolling_mean", F.mean("vwap").over(roll_w))
    .withColumn("rolling_std", F.stddev("vwap").over(roll_w))
    .withColumn(
        "z_score",
        F.when(
            F.col("rolling_std") >= MIN_STD,
            (F.col("vwap") - F.col("rolling_mean")) / F.col("rolling_std"),
        ).otherwise(F.lit(None))
    )
)

buckets.select(
    "market_slug", "side", "window_start", "vwap",
    "rolling_mean", "rolling_std", "z_score"
).orderBy("market_slug", "side", "window_start").show(10, truncate=False)
```

> **What this output means:**
>
> Three new columns have been added: `rolling_mean`, `rolling_std`, and `z_score`.
>
> - **First row** (July 24): `rolling_mean = NULL`, `z_score = NULL` — there are no preceding buckets yet so no baseline can be computed.
> - **Second row** (July 25 02:20): Still NULL — only 1 preceding bucket, need 20 to start computing properly. `rolling_std` is also NULL because you need at least 2 values to compute a standard deviation.
> - **Third row** (July 25 04:05): `rolling_mean = 0.755`, `rolling_std = 0.007`, `z_score = 2.12` — the first real z-score. The price (0.77) was 2.12 standard deviations above the recent average (0.755). This is **not** a spike — our threshold is 3.0, so this is within normal fluctuation.
>
> As more buckets accumulate, the rolling window stabilises and z-scores become more reliable.

### Add `price_before` and `price_after`

Using `lag` and `lead` window functions to pull the VWAP of the immediately adjacent buckets. These fields let us quantify how sharp the price jump was — e.g. price went from 0.08 to 0.74 in one 5-minute window.

```python
lag_w = (
    Window
    .partitionBy("market_slug", "side")
    .orderBy("window_start")
)

buckets = (
    buckets
    .withColumn("price_before", F.lag("vwap", 1).over(lag_w))
    .withColumn("price_after", F.lead("vwap", 1).over(lag_w))
)
```

## 5. Flag Spike Events

A bucket is flagged as a spike if:
1. `z_score > Z_THRESHOLD` (anomalous price move relative to baseline)
2. `volume_usd >= MIN_VOLUME` (at least $100 traded — filters single-fill artifacts in thin periods)
3. `z_score is not NULL` (warm-up buckets excluded)

```python
t0 = time.time()

spikes = (
    buckets
    .filter(F.col("volume_usd") >= MIN_VOLUME)
    .filter(F.col("z_score").isNotNull())
    .filter(F.col("z_score") > Z_THRESHOLD)
    .select(
        "market_slug",
        "side",
        "window_start",
        "window_end",
        "vwap",
        "price_before",
        "price_after",
        "rolling_mean",
        "rolling_std",
        "z_score",
        "volume_usd",
        "trade_count",
    )
    .orderBy("market_slug", "side", "window_start")
)

spike_count = spikes.count()
elapsed = time.time() - t0

print(f"Spike events detected : {spike_count}")
print(f"Z threshold           : {Z_THRESHOLD}")
print(f"Min volume (USD)      : ${MIN_VOLUME:,.0f}")
print(f"Time                  : {elapsed:.1f}s")
```

> **What this output means:**
>
> **569 spike events** were detected across all 6 markets in about 14 seconds. A spike event is a 5-minute bucket where:
> 1. The price moved more than **3 standard deviations** above the recent rolling baseline
> 2. At least **$100** was traded in that window
>
> 569 out of 65,985 total buckets = about **0.86% of all buckets** were flagged as spikes. This is expected — spikes should be rare. Most normal trading fluctuations stay within 1–2 standard deviations.

## 6. Sanity Check — Resolved Markets

For the three resolved markets we know exactly when the price spike occurred. We verify that the detector fired at the right time.

| Market | Expected spike |
|---|---|
| `us-strikes-iran-feb28` | ~2026-02-28 06:00 UTC (6.5% → 98.5% in ~1 hour) |
| `khamenei-out-feb28` | ~2026-02-28, multi-hour climb |
| `maduro-out-jan31` | ~2026-01-03 (CLOB gap — only visible in tick data) |

```python
resolved_markets = [
    ("us-strikes-iran-feb28", "2026-02-28"),
    ("khamenei-out-feb28", "2026-02-28"),
    ("maduro-out-jan31", "2026-01-03"),
]

print("=" * 70)
for slug, expected_date in resolved_markets:
    market_spikes = (
        spikes
        .filter(F.col("market_slug") == slug)
        .orderBy(F.desc("z_score"))
    )
    count = market_spikes.count()
    print(f"\n{slug}")
    print(f"  Expected resolution : {expected_date}")
    print(f"  Spikes detected     : {count}")
    if count > 0:
        market_spikes.select(
            "side",
            "window_start",
            F.round("price_before", 3).alias("before"),
            F.round("vwap", 3).alias("at_spike"),
            F.round("z_score", 2).alias("z"),
            F.round("volume_usd", 0).alias("vol_usd"),
        ).show(5, truncate=False)
    print("=" * 70)
```

> **What this output means:**
>
> This is the most important validation step. We check whether our detector found spikes on the actual resolution dates for the 3 markets where we know the answer.
>
> **US Strikes Iran — PASS ✓**
> 177 spikes detected. The resolution event (Feb 28, YES price 0.243 → 0.778, z=12.87) appears as the 2nd highest spike. The highest spike (Jan 22, z=17.89) is pre-resolution activity — someone bought YES aggressively a week before the strike.
>
> **Khamenei Out — PASS ✓**
> 63 spikes detected. The top two spikes are both on Feb 28 (the resolution date), with YES price jumping from 0.087 to 0.137 and then to 0.312 across consecutive windows. Multiple windows on the same day = a sustained price climb, consistent with news breaking and spreading.
>
> **Maduro Out — PASS ✓**
> 27 spikes detected. The single largest spike (z=33.3, price 0.12 → 0.63) lands precisely on Jan 3, 2026 — the exact resolution date. The spike on Jan 2 at 22:40 (price 0.08 → 0.14) is suspicious pre-resolution buying approximately 2.5 hours before the Jan 3 window — a strong insider signal to investigate in Phase 4.
>
> All 3 resolved markets passed. The detector correctly identifies the resolution events and also flags suspicious pre-event activity.

### Threshold Tuning Guide

If a resolved market shows **zero spikes**, lower `Z_THRESHOLD` or `MIN_VOLUME` in Cell 1 and re-run from Section 5.

If active markets show **too many spikes** (>10 each), raise `Z_THRESHOLD` to 4.0 or `MIN_VOLUME` to 500.

| Parameter | Default | Try if too few spikes | Try if too many |
|---|---|---|---|
| `Z_THRESHOLD` | 3.0 | 2.5 | 4.0 |
| `MIN_VOLUME` | 100 | 50 | 500 |
| `WINDOW_SIZE` | 20 | 10 | 30 |

## 9. Parameter Experiments

### Experiment A — Z-Score Threshold Sensitivity

How spike count changes as we raise the threshold. Goal: find the value where resolved markets still show spikes near resolution but active markets stop flooding with false positives.

```python
thresholds = [2.0, 2.5, 3.0, 3.5, 4.0, 5.0]
all_markets = [
    "us-strikes-iran-feb28", "khamenei-out-feb28", "maduro-out-jan31",
    "us-forces-iran-mar31", "us-iran-ceasefire-jun30", "china-invades-taiwan-2026"
]

print(f"{'Market':<30} {'z=2.0':>6} {'z=2.5':>6} {'z=3.0':>6} {'z=3.5':>6} {'z=4.0':>6} {'z=5.0':>6}")
print("-" * 72)

rows = {}
for slug in all_markets:
    counts = []
    for z in thresholds:
        n = (
            buckets
            .filter(F.col("market_slug") == slug)
            .filter(F.col("volume_usd") >= MIN_VOLUME)
            .filter(F.col("z_score").isNotNull())
            .filter(F.col("z_score") > z)
            .count()
        )
        counts.append(n)
    rows[slug] = counts
    short = slug[:28]
    print(f"{short:<30} {counts[0]:>6} {counts[1]:>6} {counts[2]:>6} {counts[3]:>6} {counts[4]:>6} {counts[5]:>6}")

# bar chart
import numpy as np
x = np.arange(len(all_markets))
width = 0.13
fig, ax = plt.subplots(figsize=(14, 6))
for i, (z, color) in enumerate(zip(thresholds, ["#EF5350","#FF7043","#FFA726","#66BB6A","#42A5F5","#7E57C2"])):
    vals = [rows[s][i] for s in all_markets]
    ax.bar(x + i * width, vals, width, label=f"z={z}", color=color, alpha=0.85)
ax.set_xticks(x + width * 2.5)
ax.set_xticklabels([s.replace("-", "-\n") for s in all_markets], fontsize=8)
ax.set_ylabel("Spike count")
ax.set_title("Spike Count vs Z-Score Threshold per Market")
ax.legend(title="Threshold")
ax.grid(True, axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("charts/threshold_sensitivity.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: charts/threshold_sensitivity.png")
```

> **What this output means:**
>
> This table shows how the total spike count changes for each market as we raise the z-score threshold from 2.0 to 5.0.
>
> **Key takeaways:**
>
> - At **z=2.0**, spike counts are very high (604 for US Strikes, 676 for US Forces) — too many false positives, hard to find genuine signals.
> - At **z=3.0** (our current setting), counts drop significantly but the resolution events are still captured for all 3 resolved markets.
> - At **z=4.0**, counts halve again. The main resolution spikes are still there (we saw z=33, 12, 10 for the three markets), but smaller pre-event accumulation signals get filtered out.
> - At **z=5.0**, the counts are very low. For a market like Maduro (27→8), we may be losing early pre-event signals.
>
> **Recommendation:** z=3.0 is a good starting point. If Phase 4 (bet tracing) produces too many suspect wallets to investigate, raise to z=3.5. If we miss key events, lower to z=2.5.

### Experiment B — Window Size Sensitivity

How does changing the rolling window size affect the z-score at the Maduro resolution spike? Larger window = more stable baseline, smaller = reacts faster but noisier.

```python
window_sizes = [5, 10, 20, 30, 50]
target_slug = "maduro-out-jan31"
target_ts   = "2026-01-03"

print(f"{'WINDOW_SIZE':<15} {'Z-score at resolution':<25} {'Total spikes in market'}")
print("-" * 58)

ws_results = []
for ws in window_sizes:
    w = (
        Window
        .partitionBy("market_slug", "side")
        .orderBy("window_start")
        .rowsBetween(-ws, -1)
    )
    temp = (
        buckets
        .filter(F.col("market_slug") == target_slug)
        .withColumn("rm", F.mean("vwap").over(w))
        .withColumn("rs", F.stddev("vwap").over(w))
        .withColumn(
            "zs",
            F.when(F.col("rs") >= MIN_STD,
                   (F.col("vwap") - F.col("rm")) / F.col("rs")
            ).otherwise(F.lit(None))
        )
    )
    res_row = (
        temp
        .filter(F.col("side") == "yes")
        .filter(F.col("window_start").cast("date").cast("string") == target_ts)
        .agg(F.max("zs").alias("max_z"))
        .collect()
    )
    res_z = float(res_row[0]["max_z"]) if res_row and res_row[0]["max_z"] is not None else None
    total = (
        temp
        .filter(F.col("volume_usd") >= MIN_VOLUME)
        .filter(F.col("zs").isNotNull())
        .filter(F.col("zs") > Z_THRESHOLD)
        .count()
    )
    ws_results.append((ws, res_z or 0, total))
    z_str = f"{res_z:.1f}" if res_z else "NULL"
    print(f"{ws:<15} {z_str:<25} {total}")

# chart
fig, ax1 = plt.subplots(figsize=(8, 5))
ws_vals   = [r[0] for r in ws_results]
z_vals    = [r[1] for r in ws_results]
tot_vals  = [r[2] for r in ws_results]

ax1.plot(ws_vals, z_vals, "o-", color="#2196F3", linewidth=2, label="Z-score at resolution")
ax1.set_xlabel("Window size (buckets)")
ax1.set_ylabel("Z-score at resolution spike", color="#2196F3")
ax1.tick_params(axis="y", labelcolor="#2196F3")

ax2 = ax1.twinx()
ax2.bar(ws_vals, tot_vals, width=3, color="#F44336", alpha=0.4, label="Total spikes")
ax2.set_ylabel("Total spikes detected", color="#F44336")
ax2.tick_params(axis="y", labelcolor="#F44336")

ax1.set_title(f"Window Size Sensitivity — {target_slug} YES side")
ax1.grid(True, alpha=0.3)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper right")
plt.tight_layout()
plt.savefig("charts/window_sensitivity.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: charts/window_sensitivity.png")
```

> **What this output means:**
>
> This experiment tests how changing the rolling window size affects the z-score at the Maduro resolution spike (our clearest known signal).
>
> | Window | Z at resolution | Meaning |
> |---|---|---|
> | 5 buckets (25 min) | 99.3 | Very short baseline — even a small move looks extreme |
> | 10 buckets (50 min) | 44.1 | Short baseline, high sensitivity |
> | 20 buckets (100 min) | 33.3 | **Current setting** — stable and reliable |
> | 30 buckets (150 min) | 26.8 | More history, slightly lower z |
> | 50 buckets (250 min) | 24.8 | Long baseline, adapts slowly |
>
> A smaller window makes the detector very sensitive — it picks up any deviation from recent short-term behavior. But it also means a gradual price drift over 30 minutes can make the baseline "move with" the trend, potentially missing a slow accumulation pattern.
>
> A larger window gives a more stable sense of "normal" but reacts more slowly to genuine trend changes.
>
> **The total spikes column** stays fairly consistent (19–27), meaning window size doesn't dramatically change how many things get flagged — it mainly affects how extreme the z-score looks for genuine spikes. **Window=20 is a good balance.**
